## Analysis

## Question and approach

**Question.** What does the OKN federation, taken as a whole, say about multiple sclerosis (its subtypes, genes and pathways, disease-associated expression, therapies, clinical features and biomarkers, and epidemiology), and where do the graphs agree, conflict or fall silent?

**Why it matters.** Evidence on MS is spread across curated databases, genetic association studies, expression atlases, drug resources, EHR co-occurrence and literature mining. Putting it on one disease anchor shows which mechanisms several independent kinds of evidence support, which therapies act on genetically supported targets, and where the graphs have gaps or errors.

**Approach.** The queries above anchor MS in `ubergraph` (MONDO_0005301 = DOID:2377 = UMLS C0026769) and pull evidence from 12 knowledge graphs on that anchor. The cells below map identifiers across graphs, build a gene evidence matrix that keeps evidence categories separate and ranks genes by cross-source agreement, profile expression, therapeutics, phenotypes and biomarkers, and epidemiology, and finish with enrichment, a protein-interaction network and a mechanistic map. The exact joins, thresholds and tier rules are in the specification that follows.

## Replicator specification

**Study.** Integrative multi-KG map of multiple sclerosis (MS-2026-11-19). Model `claude-opus-5-5`. OKN federated SPARQL (FRINK). Framing: hypothesis generation from curated, genetic, text-mined and observational associations; no causal or clinical inference.

**Gene evidence matrix (`analysis.evidence_matrix`).** Seven counted sources: spoke-okn, rdkg (curated); digcfdekg, wikidata P2293 (genetic); biomarkerkg analyte GID; GXA disease contrasts (E-GEOD-66573, E-GEOD-77598, E-MTAB-2973, E-MTAB-69); prokn drug→target. - **Tiers.** A = ≥ 4 sources, or ≥ 3 including curated + genetic. B = 3 sources, or 2 that are curated + genetic. C = other 2-source genes. D = 1 source.
- **Score.** n_sources + 0.25·log1p(PPI degree) + 0.05·min(PIGEAN, 10).
- **ncRNA.** Symbols matching LINC/MIR/SNHG/SNOR/RMRP/-AS are flagged and listed separately.
- **Analyst mappings.** ChEMBL complex targets → member genes; biomarkerkg analyte names (APOA1, SPHK1, SPHK2, RMRP); prokn target labels for CD52, ERVW-1 and TNFSF13B.

**Enrichment.**
- Foreground: spoke-okn ∪ rdkg genes ∪ digcfdekg PIGEAN ≥ 3 (345 symbols). Of these, 220 resolve to prokn genes; 216 carry GO BP and 183 carry Reactome annotation.
- Backgrounds: prokn genes with direct GO BP `biolink:actively_involved_in` annotation (N = 7,663), excluding NegativePropertyAssertion, or with Reactome R-HSA `participates_in` annotation (N = 6,032).
- Tested: terms with k ≥ 5 (GO, 167 terms) or k ≥ 6 (Reactome, 60 pathways).
- Statistics: hypergeometric upper tail with Benjamini–Hochberg FDR.

**Epidemiology.**
- Prevalence: IHME 2019 percent × 1,000 = per 100,000 (q13).
- Deaths: WHO GHE deaths / population × 10⁵ (q14). The `mort100k` field is not used because it is mislabeled.
- Statistics: Spearman correlations; partial ρ(latitude, prevalence | GDP) from rank residuals; Kruskal–Wallis across continents.
- GDP per capita and polygons: Natural Earth (`ne_countries.geojson` in the study `data/`), an external covariate.

**Verified quantities.**
- Evidence matrix: 1,748 genes; tiers A/B/C/D = 55/107/96/1,490.
- Enrichment: GO 159/167 and Reactome 58/60 terms at FDR < 0.05.
- Prevalence: ρ(|lat|) = 0.836 (n = 200); ρ(GDP) = 0.698 (n = 165); partial ρ = 0.743.
- Drugs: prokn has 204 MS-indicated compounds acting on 71 target genes.

**Limitations.** Associations only. 1,000-row cache cap (tails recovered). Direct, non-propagated GO annotations. GXA sex confounding (Y-chromosome genes removed). oard-kg cohort noise. biohealth predications lack effect sizes. No incidence data, sub-national prevalence or SDoH edges for MS.

### §6.2 Protein-interaction network

```python
# Figure 9 — protein–protein interaction network among the MS core signature (nestkg)
import networkx as nx
pe = A.ppi_edges(Q)
G = nx.Graph(); G.add_edges_from(zip(pe.gene_a, pe.gene_b))
tier_of = dict(zip(m.symbol, m.tier))
deg = dict(G.degree())
STATS.update({"ppi_edges": G.number_of_edges(), "ppi_nodes": G.number_of_nodes(),
              "ppi_hub1": max(deg, key=deg.get), "ppi_hub1_deg": max(deg.values())})
hubs = pd.DataFrame({"symbol": list(deg), "degree": list(deg.values())})
hubs["tier"] = hubs.symbol.map(tier_of).fillna("–")
hubs["betweenness"] = hubs.symbol.map(nx.betweenness_centrality(G))
hubs = hubs.sort_values("degree", ascending=False); hubs.to_csv(f"{EXTRACTS}/ppi_hubs.csv", index=False)
core = [n for n in G if deg[n] >= 8]
H = G.subgraph(core).copy()
H = H.subgraph(max(nx.connected_components(H), key=len)).copy()
STATS["ppi_shown_nodes"] = H.number_of_nodes()
pos = nx.spring_layout(H, seed=7, k=0.55, iterations=200)
TC = {"A": "#D55E00", "B": "#E69F00", "C": "#56B4E9", "D": "#BBBBBB", "–": "#BBBBBB"}
fig, ax = plt.subplots(figsize=(12, 10))
nx.draw_networkx_edges(H, pos, ax=ax, alpha=0.12, width=0.6)
for t, shape in [("A", "o"), ("B", "s"), ("C", "^"), ("D", "D"), ("–", "D")]:
    nl = [n for n in H if tier_of.get(n, "–") == t]
    if nl:
        nx.draw_networkx_nodes(H, pos, nodelist=nl, node_shape=shape, node_color=TC[t], ax=ax,
                               node_size=[25 + 6 * deg[n] for n in nl], edgecolors="white", linewidths=0.5,
                               label=f"tier {t}" if t != "–" else "not tiered")
lab = sorted(H, key=lambda n: -deg[n])[:30]
nx.draw_networkx_labels(H, pos, labels={n: n for n in lab}, font_size=8.5, ax=ax,
                        bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.75))
ax.axis("off")
F.legend_outside(ax, where="below", title="confidence tier (size ∝ degree)", ncol=5)
F.panel_title(ax, "", f"PPI network of MS core genes (degree ≥ 8; {H.number_of_nodes()} nodes, {H.number_of_edges()} edges)")
F.finalize(fig, 9, "figures/fig9_ppi_network.png")
```

```python
# Figure 10 — mechanistic map: MS → mechanistic module → tier A–C gene → drug (prokn drug→target edges only)
import tempfile
from mechanistic_map import render_mechanistic_map
MODULES = {
    "HLA class II / antigen presentation": ["HLA-DRB1", "HLA-DQB1", "HLA-DRA", "HLA-DPB1", "CD40", "CD86"],
    "T-cell activation & co-stimulation": ["IL2RA", "IL7R", "CD58", "CD6", "CBLB", "TAGAP", "PTPRC", "MALT1", "CD52", "DHODH"],
    "Cytokine / JAK–STAT / Th1–Th17": ["IFNG", "IL17A", "IL12A", "IL12B", "TYK2", "STAT3", "STAT4", "TNFRSF1A", "CSF2"],
    "B cell / plasma cell": ["MS4A1", "CD19", "CXCR5", "TNFSF13B", "IKZF3", "BTK"],
    "Innate / microglia / inflammasome": ["NLRP3", "IL1B", "TLR4", "ITGAM", "MERTK", "CASP1", "KIT", "CCR2"],
    "Lymphocyte trafficking / BBB": ["ITGA4", "S1PR1", "S1PR5"],
    "Redox / remyelination / conduction": ["KEAP1", "LINGO1", "RGMA", "HRH1", "KCNA1", "VDR"],
}
genes_in_kg = set(m.symbol)
assert all(g in genes_in_kg for v in MODULES.values() for g in v), [g for v in MODULES.values() for g in v if g not in genes_in_kg]
drug_short = lambda d: d.replace(" Fumaric Acid", "").replace(" Hydrochloride", "").replace(" Lauryl Sulfate", "").replace(" Fumarate", "") if "Dimethyl" not in d and "Diroximel" not in d and "Monomethyl" not in d else d
DRUGS = {}
for mod, gl in MODULES.items():
    d = tg[tg.target_gene.isin(gl) & (tg.max_phase.fillna(0) >= 2)].copy()
    d["short"] = d.drug.map(drug_short)
    d = d.sort_values(["max_phase", "short"], ascending=[False, True]).drop_duplicates("short")
    DRUGS[mod] = d.short.head(8).tolist()
mod_rows = [(mod, g, m.set_index("symbol").tier.get(g), "; ".join(sorted(set(tg[tg.target_gene == g].drug))))
            for mod, gl in MODULES.items() for g in gl]
pd.DataFrame(mod_rows, columns=["module", "gene", "tier", "prokn_drugs"]).to_csv(f"{EXTRACTS}/mechanistic_modules.csv", index=False)
STATS["mech_modules"] = len(MODULES); STATS["mech_genes"] = sum(len(v) for v in MODULES.values())
STATS["mech_drugs"] = len({d for v in DRUGS.values() for d in v})
tmp_png = render_mechanistic_map("Multiple sclerosis", MODULES, DRUGS, out_path=os.path.join(tempfile.gettempdir(), "ms_mechanistic_map_raw.png"),
                       title="MS gene–module–drug mechanistic map (entities retrieved from OKN KGs)",
                       subtitle="modules = analyst grouping of tier A–C genes guided by GO/Reactome enrichment; drugs = prokn drug→target, max phase ≥ 2",
                       gene_legend="Gene (tier A–C or drug target)", drug_legend="Drug (prokn, phase ≥ 2)")
img = plt.imread(tmp_png)  # place the rendered network on a report-style figure and save it
fig, ax = plt.subplots(figsize=(img.shape[1] / 150, img.shape[0] / 150))
ax.imshow(img); ax.axis("off")
F.finalize(fig, 10, "figures/fig10_mechanistic_map.png")
```

## Conclusions

**Summary of findings**

- **Evidence.** 1,748 genes have MS evidence in at least one graph; 55 reach tier A and 107 tier B. The top-ranked genes are IL2RA, IFNG, CD86, CD6 and IL17A.
- **Biology.** The signature is overwhelmingly immune: 159 of 167 GO BP terms and 58 of 60 Reactome pathways pass FDR < 0.05, and IFNG is the main interaction hub (degree 112 among 178 core genes).
- **Expression.** Disease expression is noisy: 11 genes recur across four MS-vs-control contrasts and 3 change in the same direction, while IFN-β treatment gives a robust interferon response.
- **Therapeutics.** prokn records 204 MS-indicated compounds acting on 71 target genes, but only IL2RA, IL17A, IL12A and VDR have tier-A genetic support, and IFN-β, glatiramer and cladribine have no drug→target edge.
- **Epidemiology.** Prevalence ranges from 1.29 to 216.5 per 100,000 across 200 countries and rises with latitude (ρ = 0.836), still 0.743 after adjusting for GDP.

**Answer.** Seen through the federation, MS is a polygenic disorder of adaptive immunity anchored on HLA class II, with an innate and microglial arm. Its therapies mostly act on lymphocyte trafficking, depletion and redox rather than on the top genetic genes, and its prevalence follows latitude independently of income. The graphs also have clear gaps and errors: no CIS term, a retained PRMS term, no NfL in biomarkerkg, and viral "causes" edges in biohealth.

**Main limitation.** Every relationship here is associative (curated, genetic, text-mined or EHR co-occurrence), and the epidemiology is ecological at country level, so nothing supports causal or clinical inference. Full limitations are in the report.
