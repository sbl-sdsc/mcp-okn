## Analysis

## Question and approach

What genes, pathways, biomarkers and drug targets explain spaceflight-induced bone loss, and how do actual spaceflight, terrestrial unloading analogs and human bone disorders agree? The approach takes vetted spaceflight-vs-ground and hindlimb-unloading contrasts from **spoke-genelab**, projects mouse genes to human orthologs, and tests the resulting signatures against curated rare low-BMD disease genes (**rdkg**) and GWAS-style BMD/osteoporosis genes (**digcfdekg**) on Entrez. Pathway programmes come from **prokn** (GO-BP and Reactome, joined on HGNC symbol), drugs from rdkg `treats` bridged through **babel** to prokn targets, and biomarkers / phenotypes from **biomarkerkg** and rdkg. Genes are integrated into an evidence score and A/B/C tiers, then synthesised into a mechanistic map. Everything is hypothesis generation.

## Replicator specification

- **Contrasts** (spoke-genelab v0.0.2; `get_valid_contrasts` rule: Space Flight arm 1 vs Ground Control arm 2, same covariates and material): OSD-690-f82a89dc (marrow, WT), OSD-690-d89cbb20 (marrow, Nrf2-KO), OSD-546-7d689d42 (hMSC, osteogenic medium). Analogs (factor arms, no factor_space): OSD-467-4d3703d7 (cortical bone, HLU vs normal loading), OSD-214-0092f94d (marrow, HLU vs loaded, no CpG / tetanus-toxoid). Log2FC > 0 = up in arm 1 (flight / unloaded).
- **DE call**: reified `MEASURED_DIFFERENTIAL_EXPRESSION_ASmMG`, adj_p_value <= 0.05 and |log2fc| >= 1. Mouse -> human via `IS_ORTHOLOG_MGiG`; human genes (taxonomy 9606) map to themselves. Many-to-one collisions keep the largest |log2FC| (q08, q09).
- **Bone gene sets**: rdkg MONDO diseases with `has_phenotype` HP_0000939 / HP_0000938 / HP_0004349, genes via `related_to` (either direction) or `genetic_association` (q12; universe N = rdkg genes with any disease link, q17). digcfdekg `geneToTrait` for 10 BMD / osteoporosis traits (q15; core = >= 3 traits; universe N = numeric-Entrez genes with any trait, q16). n = signature genes inside each universe, counted server-side (q20, q22). q21 is superseded by q22.
- **Pathways**: prokn gene `SIO_010078` -> protein -> `actively_involved_in` (GO-BP, reified, NegativePropertyAssertion removed; 10 <= K <= 500, k >= 5) or `participates_in` R-HSA (k >= 4). ## Conclusions

- The osteogenic hMSC flight signature is enriched for bone-mass genes: 113 of its genes carry >= 3 digcfdekg BMD/osteoporosis traits (2.1x, FDR 1.4e-13) and 136 are rdkg rare low-BMD disease genes (1.58x, FDR 4.0e-08); flight bone-marrow signatures show no such enrichment in the curated set.
- ALPL is the only gene down in all three flight contrasts; 9 of the 10 genes DE in all three move in opposite directions in marrow and hMSC (chemokines up in marrow, down in hMSC).
- GO-BP (63 terms) and Reactome (41 terms) at FDR < 0.05 in the hMSC signature centre on ER-to-Golgi secretion, glycosylation, UPR (effectors down), collagen / elastic fibres and osteoblast differentiation.
- 446 bone-linked flight DE genes yield 40 Tier-A genes (IL6, ENG, LMNA, FGFR2, RUNX2, COL1A1, DKK1, TNFRSF11B, ALPL, ...); none of the 8 targets of rdkg's osteoporosis drugs is DE, but their pathway neighbours are.
- 28 bone-linked DE genes are hit by phase >= 3 compounds; read with direction, CTSK (up; odanacatib) is the most coherent lead, while IL6 and PTGS2 are down in flight.

The federation supports an osteoblast / Wnt-antagonist / matrix-secretion mechanism for spaceflight bone loss with countermeasures acting through adjacent pathways; the main limitation is that only three bone-relevant flight contrasts exist and the unloading analogs in the graph are nearly empty.
